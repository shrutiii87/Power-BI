##  Naive Bayes Classifier — Python Practical


```python
from sklearn.naive_bayes import GaussianNB

# Training data
X = [[1], [2], [3], [7], [8], [9]]
y = ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass"]

# Train model
model = GaussianNB()
model.fit(X, y)

# Unseen data
new_data = [[6], [2.5], [10]]

# Prediction
predictions = model.predict(new_data)

print("Predictions:", predictions)
```

## Bernoulli Naive Bayes — Python Practical

```python

from sklearn.naive_bayes import BernoulliNB

# Training data (0 = No, 1 = Yes)
X = [[0, 0],
     [0, 1],
     [1, 0],
     [1, 1],
     [1, 1],
     [0, 0]]

y = ["Fail", "Fail", "Fail", "Pass", "Pass", "Fail"]

# Train model
model = BernoulliNB()
model.fit(X, y)

# Unseen data
new_data = [[1, 1], [0, 0], [1, 0]]

# Prediction
predictions = model.predict(new_data)

print("Predictions:", predictions)
```

## Binomial Distribution — Python Practical

```python

from scipy.stats import binom

# Given values
n = 5       # Total students
p = 0.6     # Probability of passing
k = 3       # Exactly 3 students pass

# Calculate probability
probability = binom.pmf(k, n, p)

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

## Bayes Theorem — Python Practical


```python
spam = 40
total = 100
win_in_spam = 30
total_win = 50

prior = spam / total
likelihood = win_in_spam / spam
evidence = total_win / total

posterior = (likelihood * prior) / evidence

print("Spam Probability:", posterior)
print("Percentage:", posterior * 100, "%")
```

## Conditonal Probability


```python
from collections import Counter

emails = ["Spam", "Spam", "Normal", "Spam", "Normal"]

total = len(emails)
spam = emails.count("Spam")

print("P(Spam) =", spam / total)
```
